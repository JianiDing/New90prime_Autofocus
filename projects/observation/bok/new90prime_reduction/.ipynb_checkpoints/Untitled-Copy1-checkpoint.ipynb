```python
import numpy as np
from astropy.io import fits
from astropy.stats import sigma_clipped_stats
from scipy.ndimage import median_filter

def clean_bad_pixels(image, box_size=10, sigma=4.0):
    """
    Identify and replace bad pixels in a single amplifier.
    """
    smoothed = median_filter(image, size=box_size)
    resid = image - smoothed
    mean, median, std = sigma_clipped_stats(resid)
    mask = np.abs(resid - median) > sigma * std
    #print (mask)
    cleaned = image.copy()
    cleaned[mask] = smoothed[mask]
    return cleaned

def mosaic_image(imgfile, outfile="multiamp_mosaic.fits"):
    """Create mosaic matching reference layout, with bad pixel cleaning"""
    ny, nx = imgfile[1].data.shape
    mosaic = np.zeros((2*ny, 4*nx), dtype=np.float32)

    # --- Top row (amps 5–8) ---
    mosaic[0:ny, 0:nx]     = clean_bad_pixels(imgfile[5].data)  # amp5
    mosaic[0:ny, nx:2*nx]  = clean_bad_pixels(imgfile[6].data)  # amp6
    mosaic[0:ny, 2*nx:3*nx]= clean_bad_pixels(imgfile[7].data)  # amp7
    mosaic[0:ny, 3*nx:4*nx]= clean_bad_pixels(imgfile[8].data)  # amp8

    # --- Bottom row (amps 1–4) ---
    mosaic[ny:2*ny, 0:nx]     = clean_bad_pixels(imgfile[1].data)  # amp1
    mosaic[ny:2*ny, nx:2*nx]  = clean_bad_pixels(imgfile[2].data)  # amp2
    mosaic[ny:2*ny, 2*nx:3*nx]= clean_bad_pixels(imgfile[3].data)  # amp3
    mosaic[ny:2*ny, 3*nx:4*nx]= clean_bad_pixels(imgfile[4].data)  # amp4

    fits.writeto(outfile, mosaic, overwrite=True)
    return mosaic
```

```python
import sep
finalcut = np.ascontiguousarray(final[0:2500,0:1200].astype(np.float32))
bkg = sep.Background(finalcut)
```

```python
plt.imshow(finalcut,vmin=0,vmax=20)
```

```python
# get a "global" mean and noise of the image background:
print(bkg.globalback)
print(bkg.globalrms)
```

```python
# evaluate background as 2-d array, same size as original image
bkg_image = bkg.back()
plt.imshow(bkg_image, interpolation='nearest', cmap='gray', origin='lower',vmax=20)
plt.colorbar()
```

```python
# evaluate the background noise as 2-d array, same size as original image
bkg_rms = bkg.rms()
# show the background noise
plt.imshow(bkg_rms, interpolation='nearest', cmap='gray', origin='lower',vmin=0,vmax=50)
plt.colorbar()
```

```python
data_sub = finalcut - bkg
```

```python
plt.imshow(data_sub,vmin=0,vmax=20)
```

```python
data_sub[0:4000,0:1100]
```

```python
objects = sep.extract(data_sub, 5, err=bkg.globalrms)
```

```python
from matplotlib.patches import Ellipse
plt.figure(figsize=(18, 10), dpi=100)
# plot background-subtracted image
fig, ax = plt.subplots()
m, s = np.mean(data_sub), np.std(data_sub)
im = ax.imshow(data_sub, interpolation='nearest', cmap='gray',
               vmin=0, vmax=200, origin='lower')

# plot an ellipse for each object
for i in range(len(objects)):
    e = Ellipse(xy=(objects['x'][i], objects['y'][i]),
                width=10*objects['a'][i],
                height=10*objects['b'][i],
                angle=objects['theta'][i] * 180. / np.pi)
    e.set_facecolor('none')
    e.set_edgecolor('red')
    ax.add_artist(e)
```

```python
from astropy.io import fits
import numpy as np

# segmentation map: 0 = background, i+1 = object index
segmap = np.zeros_like(data_sub, dtype=np.int16)
for i in range(len(objects)):
    x, y = int(objects['x'][i]), int(objects['y'][i])
    if 0 <= y < segmap.shape[0] and 0 <= x < segmap.shape[1]:
        segmap[y, x] = i+1

# make HDUs
hdu_image = fits.PrimaryHDU(data_sub.astype(np.float32))
hdu_seg = fits.ImageHDU(segmap, name="SEGMENTATION")

hdul = fits.HDUList([hdu_image, hdu_seg])
hdul.writeto("detections.fits", overwrite=True)
```

```python
with open("detections.reg", "w") as f:
    f.write("# Region file format: DS9 version 4.1\n")
    f.write("image\n")
    for i in range(len(objects)):
        x = objects['x'][i]
        y = objects['y'][i]
        a = 3 * objects['a'][i]
        b = 3 * objects['b'][i]
        theta = objects['theta'][i] * 180. / np.pi
        f.write(f"ellipse({x},{y},{a},{b},{theta})\n")
```
